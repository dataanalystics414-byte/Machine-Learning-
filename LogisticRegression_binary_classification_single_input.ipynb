```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
dataset = pd.read_csv(r"C:\Users\AMAN VIRKHARE\sql project\Pyhton_Project\DATA SCIENCE\age_salary_purchase_dataset.csv")
dataset.drop(columns = ['EstimatedSalary'], inplace = True ) 
```

```python
dataset.head(5)
```

```python
plt.figure(figsize  = (4, 3))

sns.scatterplot(x = 'Age', y = 'Purchased', data = dataset) 
plt.show()
```

```python
x = dataset[['Age']]
y = dataset['Purchased']
```

```python
from sklearn.model_selection import train_test_split 
```

```python
x_train , x_test , y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr= LogisticRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)
```

```python
lr.predict([[30]])
```

```python

plt.figure(figsize=(5, 4))

# 1. Remove the quotes around dataset
sns.scatterplot(x="Age", y="Purchased", data=dataset)

# 2. Pass a 2D DataFrame to predict(), and flatten the output to 1D
sns.lineplot(x="Age", y=lr.predict(dataset[["Age"]]).ravel(), data=dataset, color="red")

plt.show()
```
